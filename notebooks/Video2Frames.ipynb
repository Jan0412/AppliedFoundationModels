```python
from pathlib import Path
import cv2

def extract_native_frames(video_path: str, output_dir: str, img_format: str = "jpg") -> None:
    """
    Extracts all frames from a video file at its native FPS and saves them to disk.
    """
    vid_path = Path(video_path)
    out_dir = Path(output_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    cap = cv2.VideoCapture(str(vid_path))
    if not cap.isOpened():
        raise IOError(f"Cannot open video stream: {vid_path}")

    native_fps = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    
    print(f"Target: {vid_path.name}")
    print(f"Native FPS: {native_fps:.2f} | Total Frames: {total_frames}")

    frame_idx = 0
    while True:
        success, frame = cap.read()
        if not success:
            break
            
        out_file = out_dir / f"{frame_idx}.{img_format}"
        cv2.imwrite(str(out_file), frame)
        frame_idx += 1

    cap.release()
    print(f"Completed: Saved {frame_idx} frames to {out_dir}/")

# Usage
# extract_native_frames(video_path="iphone_video.MOV", output_dir="./iphone_frames")
```
